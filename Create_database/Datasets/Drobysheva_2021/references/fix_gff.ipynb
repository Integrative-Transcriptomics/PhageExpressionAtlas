# Rewrite gff for processing with pipeline

```python
import pandas as pd
```

```python
# Read gff file

gff = pd.read_csv('MGRFG5_1_Cba_WT_1_reference.gff3', sep='\t', comment='#', header=None)
gff = gff.loc[gff[0] =='contig_1']
gff
```

```python
gff[8] = gff[8] + ';gene_biotype=' + gff[2]
gff
```

```python
set(gff[2].tolist()) # replace all types with gene except for region and regulatory region
```

```python
gff.replace(['CDS', 'ncRNA', 'rRNA', 'tRNA', 'tmRNA'], 'gene', inplace=True)
gff
```

```python
# Write to output

gff.to_csv('MGRFG5_1_Cba_WT_1_reference_adjusted.gff3', sep='\t', header=False, index=False)
```
